## 1. Problem Statement

### Physical Setup

A reinforced concrete column is exposed to fire on its surface. We need to determine:
- Temperature distribution through the column thickness as a function of time
- Time until steel reinforcement reaches critical temperature (538°C)
- Fire resistance rating for building code compliance

### Assumptions

1. **One-dimensional heat transfer** (perpendicular to surface)
2. **Homogeneous material properties** (constant thermal properties)
3. **No internal heat generation**
4. **Column is symmetric** (analyze half-width only)

---

## 2. Governing Equation

### Heat Diffusion Equation

The temperature $T(x,t)$ in the concrete is governed by the **1D heat diffusion equation**:

$$\frac{\partial T}{\partial t} = \alpha \frac{\partial^2 T}{\partial x^2}$$

where:
- $T(x,t)$ = temperature [K] at position $x$ and time $t$
- $\alpha$ = thermal diffusivity [m²/s] = $\frac{k}{\rho c_p}$
- $k$ = thermal conductivity [W/m·K]
- $\rho$ = density [kg/m³]
- $c_p$ = specific heat capacity [J/kg·K]
- $x$ = distance from column centerline [m]
- $t$ = time [s]

### Physical Interpretation

The equation states:
> **Rate of temperature change** = **Thermal diffusivity** × **Curvature of temperature profile**

- Regions with high curvature (steep temperature gradients) → rapid temperature change
- Thermal diffusivity $\alpha$ controls how fast heat propagates through the material
- For concrete: $\alpha \approx 5.8 \times 10^{-7}$ m²/s (heat moves slowly)

---

## 3. Initial and Boundary Conditions

### Initial Condition

At $t = 0$, the entire column is at ambient temperature:

$$T(x, 0) = T_{\text{ambient}} = 20°\text{C} = 293 \text{ K}$$

### Boundary Conditions

**At the centerline** ($x = 0$): **Symmetry condition**

$$\left. \frac{\partial T}{\partial x} \right|_{x=0} = 0$$

No heat flux crosses the centerline (by symmetry).

**At the surface** ($x = L$): **Prescribed temperature (fire exposure)**

$$T(L, t) = T_{\text{fire}}(t)$$

where $T_{\text{fire}}(t)$ follows the **ASTM E119 standard fire curve**:

$$T_{\text{fire}}(t) = T_{\text{ambient}} + 750 \left(1 - e^{-3.79553\sqrt{t_{\text{min}}}} \right)$$

with $t_{\text{min}}$ in minutes.

---

## 4. Finite Difference Discretization

### Spatial Discretization

Divide the domain $[0, L]$ into $n$ equally-spaced nodes:

$$x_i = i \cdot \Delta x, \quad i = 0, 1, 2, \ldots, n-1$$

where $\Delta x = \frac{L}{n-1}$ is the spatial step size.

### Temporal Discretization

Divide time into discrete steps:

$$t^n = n \cdot \Delta t, \quad n = 0, 1, 2, \ldots, N$$

where $\Delta t$ is the time step size.

### Notation

Let $T_i^n$ denote the temperature at spatial node $i$ and time level $n$:

$$T_i^n \approx T(x_i, t^n)$$

---

## 5. Finite Difference Approximations

### Forward Difference in Time (1st Order)

The time derivative is approximated using a **forward difference** (explicit):

$$\frac{\partial T}{\partial t} \bigg|_i^n \approx \frac{T_i^{n+1} - T_i^n}{\Delta t}$$

**Truncation error:** $O(\Delta t)$ (first-order accurate)

**Derivation:** Taylor series expansion:

$$T_i^{n+1} = T_i^n + \Delta t \frac{\partial T}{\partial t}\bigg|_i^n + \frac{(\Delta t)^2}{2} \frac{\partial^2 T}{\partial t^2}\bigg|_i^n + O(\Delta t^3)$$

Rearranging:

$$\frac{\partial T}{\partial t}\bigg|_i^n = \frac{T_i^{n+1} - T_i^n}{\Delta t} - \frac{\Delta t}{2} \frac{\partial^2 T}{\partial t^2}\bigg|_i^n + O(\Delta t^2)$$

The second term is the truncation error $O(\Delta t)$.

### Central Difference in Space (2nd Order)

The second spatial derivative is approximated using a **central difference**:

$$\frac{\partial^2 T}{\partial x^2} \bigg|_i^n \approx \frac{T_{i+1}^n - 2T_i^n + T_{i-1}^n}{(\Delta x)^2}$$

**Truncation error:** $O(\Delta x^2)$ (second-order accurate)

**Derivation:** Taylor series expansions:

Forward:
$$T_{i+1}^n = T_i^n + \Delta x \frac{\partial T}{\partial x}\bigg|_i^n + \frac{(\Delta x)^2}{2} \frac{\partial^2 T}{\partial x^2}\bigg|_i^n + \frac{(\Delta x)^3}{6} \frac{\partial^3 T}{\partial x^3}\bigg|_i^n + O(\Delta x^4)$$

Backward:
$$T_{i-1}^n = T_i^n - \Delta x \frac{\partial T}{\partial x}\bigg|_i^n + \frac{(\Delta x)^2}{2} \frac{\partial^2 T}{\partial x^2}\bigg|_i^n - \frac{(\Delta x)^3}{6} \frac{\partial^3 T}{\partial x^3}\bigg|_i^n + O(\Delta x^4)$$

Adding these two equations:

$$T_{i+1}^n + T_{i-1}^n = 2T_i^n + (\Delta x)^2 \frac{\partial^2 T}{\partial x^2}\bigg|_i^n + O(\Delta x^4)$$

Rearranging:

$$\frac{\partial^2 T}{\partial x^2}\bigg|_i^n = \frac{T_{i+1}^n - 2T_i^n + T_{i-1}^n}{(\Delta x)^2} + O(\Delta x^2)$$

Note: The first and third derivative terms cancel out, giving **second-order accuracy**.

---

## 6. Explicit Finite Difference Scheme

### Discretized Equation

Substituting the finite difference approximations into the heat equation:

$$\frac{T_i^{n+1} - T_i^n}{\Delta t} = \alpha \frac{T_{i+1}^n - 2T_i^n + T_{i-1}^n}{(\Delta x)^2}$$

### Solution Formula

Solving for $T_i^{n+1}$ (the temperature at the next time step):

$$T_i^{n+1} = T_i^n + \frac{\alpha \Delta t}{(\Delta x)^2} \left( T_{i+1}^n - 2T_i^n + T_{i-1}^n \right)$$

Define the **Fourier number** (or **mesh Fourier number**):

$$r = \frac{\alpha \Delta t}{(\Delta x)^2}$$

Then the scheme simplifies to:

$$\boxed{T_i^{n+1} = T_i^n + r \left( T_{i+1}^n - 2T_i^n + T_{i-1}^n \right)}$$

or equivalently:

$$\boxed{T_i^{n+1} = r T_{i-1}^n + (1 - 2r) T_i^n + r T_{i+1}^n}$$

### Why "Explicit"?

This method is called **explicit** because $T_i^{n+1}$ is computed directly from known values at time level $n$:
- No system of equations to solve
- Easy to implement
- Fast computation per time step
- **But**: conditionally stable (see below)

---

## 7. Boundary Condition Implementation

### Centerline (Symmetry): $i = 0$

From the symmetry condition $\frac{\partial T}{\partial x}\big|_{x=0} = 0$, we use a **one-sided difference**:

$$\frac{T_1^n - T_{-1}^n}{2\Delta x} = 0$$

This gives: $T_{-1}^n = T_1^n$ (ghost point)

Substituting into the scheme at $i=0$:

$$T_0^{n+1} = T_0^n + r(T_1^n - 2T_0^n + T_{-1}^n) = T_0^n + r(T_1^n - 2T_0^n + T_1^n)$$

$$\boxed{T_0^{n+1} = T_0^n + 2r(T_1^n - T_0^n)}$$

**Simplified implementation:** Since symmetry implies no gradient, we can also use:

$$\boxed{T_0^{n+1} = T_1^{n+1}}$$

### Surface (Fire Exposure): $i = n-1$

Prescribed temperature from fire curve:

$$\boxed{T_{n-1}^{n+1} = T_{\text{fire}}(t^{n+1})}$$

### Interior Nodes: $i = 1, 2, \ldots, n-2$

Use the standard explicit formula:

$$\boxed{T_i^{n+1} = r T_{i-1}^n + (1 - 2r) T_i^n + r T_{i+1}^n}$$

---

## 8. Stability Analysis

### Von Neumann Stability Analysis

For the explicit scheme to be **stable**, numerical errors must not grow exponentially. The stability criterion is:

$$\boxed{r = \frac{\alpha \Delta t}{(\Delta x)^2} \leq \frac{1}{2}}$$

**Physical interpretation:**
- Each spatial node receives heat from neighbors weighted by $r$
- The center node coefficient is $(1-2r)$
- For stability: $(1-2r) \geq 0 \Rightarrow r \leq 0.5$
- If $r > 0.5$: numerical oscillations grow without bound

### Maximum Stable Time Step

Rearranging the stability criterion:

$$\Delta t \leq \frac{(\Delta x)^2}{2\alpha}$$

**Example:** For concrete with $\alpha = 5.8 \times 10^{-7}$ m²/s and $\Delta x = 0.004$ m:

$$\Delta t \leq \frac{(0.004)^2}{2 \times 5.8 \times 10^{-7}} = 13.8 \text{ s}$$

Using $\Delta t = 1.08$ s gives $r = 0.039 \ll 0.5$ ✓ **Very stable**

### What Happens if Unstable?

If you violate the stability criterion ($r > 0.5$):
- The coefficient $(1-2r)$ becomes **negative**
- Solution exhibits unphysical oscillations
- Errors grow exponentially
- Solution "blows up" (temperatures → ±∞)

This is a great demonstration for students!

## 9. Algorithm Summary

### Initialization

```
1. Set initial condition: T[i, 0] = T_ambient for all i
2. Calculate r = α·Δt/(Δx)²
3. Check stability: if r > 0.5, reduce Δt or increase Δx
```

### Time-Stepping Loop

```
For n = 0, 1, 2, ..., N-1:
    
    // Update fire temperature
    T_fire = fire_temperature(t^n)
    
    // Interior nodes (explicit formula)
    For i = 1, 2, ..., n-2:
        T[i, n+1] = T[i, n] + r·(T[i+1, n] - 2·T[i, n] + T[i-1, n])
    
    // Boundary: centerline (symmetry)
    T[0, n+1] = T[1, n+1]
    
    // Boundary: surface (fire exposure)
    T[n-1, n+1] = T_fire
    
    // Check rebar temperature
    If T[rebar_index, n+1] ≥ T_critical:
        Record failure time and exit
```

## 10. Engineering Design Considerations

### Fire Resistance Rating

Building codes require minimum fire resistance:
- **1-hour rating:** Most commercial buildings
- **2-hour rating:** High-rise buildings, exit stairwells
- **3-hour rating:** Critical infrastructure

**Criterion:** Steel reinforcement must stay below 538°C (1000°F)
- Steel loses 50% of its yield strength at this temperature
- Structural failure becomes likely

### Concrete Cover Thickness

The distance from rebar to fire-exposed surface:
- **More cover** → better fire protection, higher rating
- **Less cover** → lower cost, more usable space

**Typical values:**
- 2 inches (5 cm): Standard for columns
- 3 inches (7.5 cm): Enhanced fire protection
- 1.5 inches (3.8 cm): Minimum for corrosion protection

### Material Properties (Temperature-Dependent)

**Advanced models** account for:
- Thermal conductivity $k(T)$ decreases at high temperatures
- Specific heat $c_p(T)$ increases (moisture evaporation)
- Concrete spalling at high temperatures
- Radiation and convection at the surface

Our simplified model assumes constant properties (conservative).

## 11. Validation

### Analytical Solution (Constant Surface Temperature)

For constant $T_{\text{surface}}$ applied at $t=0$, the analytical solution is:

$$T(x,t) = T_{\text{surface}} + (T_{\text{initial}} - T_{\text{surface}}) \cdot \text{erf}\left(\frac{x}{2\sqrt{\alpha t}}\right)$$

where $\text{erf}(z)$ is the error function.

**Use:** Validate numerical code against this solution before applying time-varying fire curve.

### Expected Behavior

- Surface temperature follows fire curve immediately
- Interior temperature lags (thermal inertia)
- Temperature gradient is steepest near surface
- Heat penetration depth $\sim \sqrt{\alpha t}$ (diffusive scaling)

---

## 12. Computational Considerations

### Mesh Sensitivity

**Spatial mesh:**
- Finer mesh → better accuracy, smaller stable $\Delta t$
- Rule of thumb: $\Delta x \approx$ 1-5 mm for concrete

**Temporal mesh:**
- Limited by stability criterion
- Typical: $\Delta t = 1-10$ seconds

### Efficiency

For explicit method:
- **Cost per time step:** $O(n)$ (loop over nodes)
- **Total cost:** $O(n \times N)$ where $N$ = number of time steps
- Very fast for 1D problems
- Memory: store only two time levels

### Sparse Matrix Structure

If written in matrix form: $\mathbf{T}^{n+1} = \mathbf{A} \mathbf{T}^n + \mathbf{b}^n$

Matrix $\mathbf{A}$ is **tridiagonal** (only 3 non-zero entries per row).

---

## 13. Key Takeaways for Students

1. **Forward difference in time** → explicit method → conditionally stable
2. **Central difference in space** → second-order accurate
3. **Stability criterion** $r \leq 0.5$ is **critical** – violate it and solution fails
4. Explicit methods are simple but require small time steps
5. Physical interpretation: heat diffuses slowly through concrete (protective effect)
6. Engineering trade-off: cover thickness vs. cost vs. fire rating
7. Building codes specify minimum fire resistance based on occupancy
8. Numerical methods handle time-varying boundary conditions easily

````python
"""
CE311K - Computer Methods for Civil Engineering
Example: Heat Conduction in Concrete Column During Fire

Problem: Structural fire safety analysis
How long before a steel reinforcement bar reaches critical temperature?

Governing Equation: ∂T/∂t = α * ∂²T/∂x²  (1D heat diffusion)

Time discretization: Forward difference (explicit)
Space discretization: Central difference

Application: Fire resistance rating of reinforced concrete columns
"""

import numpy as np
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation

# ============================================================================
# PROBLEM SETUP: CONCRETE COLUMN IN FIRE
# ============================================================================

print("="*70)
print("STRUCTURAL FIRE SAFETY: CONCRETE COLUMN HEAT TRANSFER")
print("Explicit Finite Difference Method")
print("="*70)

# Column geometry
width = 0.40  # Column width (m) - typical 16" column
cover = 0.05  # Concrete cover to rebar center (m) - 2" cover

# Concrete thermal properties
rho = 2400  # Density (kg/m³)
c_p = 1000  # Specific heat (J/kg·K)
k_concrete = 1.4  # Thermal conductivity (W/m·K)
alpha = k_concrete / (rho * c_p)  # Thermal diffusivity (m²/s)

# Steel rebar properties (for critical temperature check)
T_critical_steel = 538 + 273  # Critical temp for steel = 538°C (1000°F) in Kelvin

# Fire exposure (Standard fire curve - ASTM E119 / ISO 834)
T_ambient = 20 + 273  # Initial temperature (K)
def fire_temperature(t_minutes):
    """ASTM E119 standard fire curve: T = T_0 + 750*(1 - e^(-3.79553*sqrt(t)))"""
    if t_minutes <= 0:
        return T_ambient
    T_fire_celsius = 20 + 750 * (1 - np.exp(-3.79553 * np.sqrt(t_minutes)))
    return T_fire_celsius + 273  # Convert to Kelvin

# Simulation parameters
t_total = 3 * 3600  # 3 hours in seconds (typical fire rating requirement)
n_time_steps = 10000
dt = t_total / n_time_steps

# Spatial discretization (half-width due to symmetry)
x = np.linspace(0, width/2, 51)  # From center to surface
n_nodes = len(x)
dx = x[1] - x[0]

# Stability check
r = alpha * dt / (dx**2)

print(f"\nColumn Geometry:")
print(f"  Width: {width*100:.0f} cm ({width*39.37:.0f} inches)")
print(f"  Rebar cover: {cover*100:.1f} cm ({cover*39.37:.1f} inches)")
print(f"  Rebar location: {cover:.3f} m from surface")

print(f"\nMaterial Properties (Concrete):")
print(f"  Thermal diffusivity: {alpha:.2e} m²/s")
print(f"  Thermal conductivity: {k_concrete} W/m·K")

print(f"\nFire Exposure:")
print(f"  Standard: ASTM E119")
print(f"  Duration: {t_total/3600:.0f} hours")
print(f"  Initial temp: {T_ambient-273:.0f}°C")
print(f"  Fire temp (1hr): {fire_temperature(60)-273:.0f}°C")
print(f"  Fire temp (2hr): {fire_temperature(120)-273:.0f}°C")
print(f"  Steel critical temp: {T_critical_steel-273:.0f}°C")

print(f"\nNumerical Parameters:")
print(f"  Spatial nodes: {n_nodes}")
print(f"  Time steps: {n_time_steps}")
print(f"  Δx = {dx*100:.2f} cm")
print(f"  Δt = {dt:.2f} s = {dt/60:.3f} minutes")

print(f"\nStability Check:")
print(f"  r = α*Δt/Δx² = {r:.4f}")
if r <= 0.5:
    print(f"  ✓ STABLE (r ≤ 0.5)")
else:
    print(f"  ✗ UNSTABLE (r > 0.5)")
    print(f"  Required: Δt ≤ {0.5*dx**2/alpha:.2f} s")
    exit()

# ============================================================================
# ANALYTICAL SOLUTION (for validation)
# ============================================================================
"""
For constant surface temperature (simplified case), analytical solution exists:
T(x,t) = T_surface + (T_initial - T_surface) * erf(x/(2*sqrt(α*t)))

We'll use this to validate our numerical method at a fixed fire temperature
"""

def analytical_solution_const_temp(x_vals, t, T_surface, T_initial, alpha):
    """Semi-infinite solid with constant surface temperature"""
    from scipy.special import erf
    if t == 0:
        return np.full_like(x_vals, T_initial)
    return T_surface + (T_initial - T_surface) * erf(x_vals / (2*np.sqrt(alpha*t)))

# ============================================================================
# EXPLICIT FINITE DIFFERENCE SOLUTION
# ============================================================================
"""
Governing equation: ∂T/∂t = α * ∂²T/∂x²

Forward difference in time:
(T[i,n+1] - T[i,n])/Δt = α * ∂²T/∂x²|[i,n]

Central difference in space:
∂²T/∂x²|[i,n] ≈ (T[i+1,n] - 2*T[i,n] + T[i-1,n])/Δx²

Explicit scheme:
T[i,n+1] = T[i,n] + r*(T[i+1,n] - 2*T[i,n] + T[i-1,n])

where r = α*Δt/Δx²

Boundary conditions:
- x=0 (center): ∂T/∂x = 0 (symmetry) → T[0] = T[1]
- x=width/2 (surface): T = T_fire(t)
"""

# Initialize arrays
T = np.zeros((n_nodes, n_time_steps + 1))
T[:, 0] = T_ambient  # Initial condition

# Track time and rebar temperature
time = np.linspace(0, t_total, n_time_steps + 1)
time_minutes = time / 60

# Find rebar location index
rebar_index = np.argmin(np.abs(x - (width/2 - cover)))
rebar_distance = x[rebar_index]

print(f"\n" + "="*70)
print("SOLVING HEAT TRANSFER...")
print("="*70)

# Store snapshots for visualization
snapshot_times_min = [0, 5, 15, 30, 60, 120, 180]  # minutes
snapshot_indices = []
for t_snap in snapshot_times_min:
    idx = np.argmin(np.abs(time_minutes - t_snap))
    snapshot_indices.append(idx)

# Time stepping
rebar_critical_time = None

for n in range(n_time_steps):
    # Update fire temperature
    T_fire = fire_temperature(time_minutes[n])

    # Interior nodes (explicit scheme)
    for i in range(1, n_nodes - 1):
        T[i, n+1] = T[i, n] + r * (T[i+1, n] - 2*T[i, n] + T[i-1, n])

    # Boundary condition at center (symmetry)
    T[0, n+1] = T[1, n+1]

    # Boundary condition at surface (fire exposure)
    T[-1, n+1] = T_fire

    # Check if rebar reached critical temperature
    if rebar_critical_time is None and T[rebar_index, n+1] >= T_critical_steel:
        rebar_critical_time = time[n+1] / 60  # in minutes

    # Progress indicator
    if (n+1) % 1000 == 0:
        progress = (n+1) / n_time_steps * 100
        print(f"  Progress: {progress:.0f}% - Time: {time_minutes[n]:.1f} min - "
              f"Surface: {T[-1,n]-273:.0f}°C - Rebar: {T[rebar_index,n]-273:.0f}°C")

print("✓ Solution complete!")

# ============================================================================
# RESULTS ANALYSIS
# ============================================================================

print(f"\n" + "="*70)
print("FIRE RESISTANCE ANALYSIS")
print("="*70)

# Final temperatures
T_surface_final = T[-1, -1] - 273
T_center_final = T[0, -1] - 273
T_rebar_final = T[rebar_index, -1] - 273

print(f"\nFinal Temperatures (after {t_total/3600:.0f} hours):")
print(f"  Surface (fire-exposed): {T_surface_final:.0f}°C")
print(f"  Rebar location: {T_rebar_final:.0f}°C")
print(f"  Column center: {T_center_final:.0f}°C")

print(f"\nSteel Reinforcement Analysis:")
print(f"  Critical temperature: {T_critical_steel-273:.0f}°C")
print(f"  Rebar location: {rebar_distance*100:.1f} cm from center ({(width/2-rebar_distance)*100:.1f} cm cover)")

if rebar_critical_time:
    print(f"  ✗ CRITICAL: Rebar reached {T_critical_steel-273:.0f}°C after {rebar_critical_time:.1f} minutes")
    print(f"  Fire rating: < {rebar_critical_time:.0f} minutes")
    if rebar_critical_time < 60:
        print(f"  RECOMMENDATION: Increase cover or use fire protection")
    elif rebar_critical_time < 120:
        print(f"  RECOMMENDATION: Adequate for 1-hour rating, consider upgrade for 2-hour")
    else:
        print(f"  Rating: Adequate for 2-hour fire resistance")
else:
    print(f"  ✓ SAFE: Rebar stayed below {T_critical_steel-273:.0f}°C for entire {t_total/3600:.0f} hour test")
    print(f"  Fire rating: > {t_total/60:.0f} minutes")

# ============================================================================
# VISUALIZATION
# ============================================================================

fig = plt.figure(figsize=(15, 10))

# Plot 1: Temperature profiles at different times
ax1 = plt.subplot(2, 2, 1)
colors = plt.cm.YlOrRd(np.linspace(0.3, 1, len(snapshot_indices)))
for idx, snap_idx in enumerate(snapshot_indices):
    t_min = time_minutes[snap_idx]
    ax1.plot(x*100, T[:, snap_idx]-273, 'o-', linewidth=2, markersize=4,
             color=colors[idx], label=f't = {t_min:.0f} min', markevery=3)

ax1.axhline(T_critical_steel-273, color='red', linestyle='--', linewidth=2,
            label=f'Critical temp ({T_critical_steel-273:.0f}°C)')
ax1.axvline(rebar_distance*100, color='blue', linestyle=':', linewidth=2,
            label=f'Rebar location')
ax1.set_xlabel('Distance from Column Center (cm)', fontsize=11)
ax1.set_ylabel('Temperature (°C)', fontsize=11)
ax1.set_title('Temperature Profiles Through Column Thickness', fontsize=13, fontweight='bold')
ax1.grid(True, alpha=0.3)
ax1.legend(fontsize=9, loc='upper left')
ax1.set_xlim([0, width/2*100])

# Plot 2: Temperature history at key locations
ax2 = plt.subplot(2, 2, 2)
ax2.plot(time_minutes, T[-1, :]-273, 'r-', linewidth=2.5, label='Surface (fire-exposed)')
ax2.plot(time_minutes, T[rebar_index, :]-273, 'b-', linewidth=2.5,
         label=f'Rebar ({(width/2-rebar_distance)*100:.1f} cm cover)')
ax2.plot(time_minutes, T[0, :]-273, 'g-', linewidth=2.5, label='Column center')
ax2.axhline(T_critical_steel-273, color='orange', linestyle='--', linewidth=2,
            label=f'Critical temp ({T_critical_steel-273:.0f}°C)')
if rebar_critical_time:
    ax2.axvline(rebar_critical_time, color='red', linestyle=':', linewidth=2,
                label=f'Failure time ({rebar_critical_time:.0f} min)')
ax2.set_xlabel('Time (minutes)', fontsize=11)
ax2.set_ylabel('Temperature (°C)', fontsize=11)
ax2.set_title('Temperature History During Fire Exposure', fontsize=13, fontweight='bold')
ax2.grid(True, alpha=0.3)
ax2.legend(fontsize=9)
ax2.set_xlim([0, t_total/60])

# Plot 3: Fire curve comparison
ax3 = plt.subplot(2, 2, 3)
t_fire_test = np.linspace(0, 180, 100)
T_fire_curve = [fire_temperature(t)-273 for t in t_fire_test]
ax3.plot(t_fire_test, T_fire_curve, 'r-', linewidth=3, label='ASTM E119 Standard Fire')
ax3.plot(time_minutes, T[-1, :]-273, 'b--', linewidth=2, label='Column surface temp')
ax3.set_xlabel('Time (minutes)', fontsize=11)
ax3.set_ylabel('Temperature (°C)', fontsize=11)
ax3.set_title('Standard Fire Curve vs Surface Temperature', fontsize=13, fontweight='bold')
ax3.grid(True, alpha=0.3)
ax3.legend(fontsize=10)
ax3.set_xlim([0, 180])

# Plot 4: Heatmap of temperature evolution
ax4 = plt.subplot(2, 2, 4)
# Downsample for visualization
time_downsample = time_minutes[::20]
T_downsample = T[:, ::20] - 273
X, Y = np.meshgrid(time_downsample, x*100)
contour = ax4.contourf(X, Y, T_downsample, levels=20, cmap='hot')
contour_lines = ax4.contour(X, Y, T_downsample, levels=[100, 200, 300, 400, 538],
                             colors='white', linewidths=1, linestyles='--')
ax4.clabel(contour_lines, inline=True, fontsize=8, fmt='%d°C')
ax4.axhline(rebar_distance*100, color='cyan', linestyle='-', linewidth=2,
            label='Rebar location')
cbar = plt.colorbar(contour, ax=ax4)
cbar.set_label('Temperature (°C)', fontsize=10)
ax4.set_xlabel('Time (minutes)', fontsize=11)
ax4.set_ylabel('Distance from Center (cm)', fontsize=11)
ax4.set_title('Temperature Evolution (Space-Time)', fontsize=13, fontweight='bold')
ax4.legend(fontsize=9, loc='lower right')

plt.tight_layout()
plt.savefig('fire_resistance_analysis.png', dpi=300, bbox_inches='tight')
print(f"\n✓ Analysis plots saved!")

# ============================================================================
# VALIDATION AGAINST ANALYTICAL SOLUTION (constant temperature case)
# ============================================================================

# For validation, compare against analytical solution at a fixed temperature
print(f"\n" + "="*70)
print("VALIDATION (Constant Surface Temp = 1000°C)")
print("="*70)

T_fixed = 1000 + 273  # Fixed fire temperature for validation
t_validate = 60 * 60  # 1 hour
n_validate = int(t_validate / dt)

# Numerical solution with fixed surface temp
T_validate_num = np.zeros((n_nodes, n_validate + 1))
T_validate_num[:, 0] = T_ambient

for n in range(n_validate):
    for i in range(1, n_nodes - 1):
        T_validate_num[i, n+1] = T_validate_num[i, n] + r * (
            T_validate_num[i+1, n] - 2*T_validate_num[i, n] + T_validate_num[i-1, n]
        )
    T_validate_num[0, n+1] = T_validate_num[1, n+1]
    T_validate_num[-1, n+1] = T_fixed

# Analytical solution
T_validate_analytical = analytical_solution_const_temp(x, t_validate, T_fixed, T_ambient, alpha)

# Calculate error
error = np.abs(T_validate_num[:, -1] - T_validate_analytical)
max_error = np.max(error)
rms_error = np.sqrt(np.mean(error**2))

print(f"After {t_validate/3600:.0f} hour with constant surface temp:")
print(f"  Maximum error: {max_error:.2f} K")
print(f"  RMS error: {rms_error:.2f} K")
print(f"  Relative error: {max_error/(T_fixed-T_ambient)*100:.2f}%")
if max_error < 5:
    print(f"  ✓ Excellent agreement with analytical solution")

plt.show()

print(f"\n" + "="*70)
print("Analysis complete! Check the plots for detailed results.")
print("="*70)
````
